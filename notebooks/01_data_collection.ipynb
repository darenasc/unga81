```python
from unga81.data_collection import (
    get_video_url_list,
    download_all_transcripts,
    get_corpus_from_file,
    get_seconds_from_str,
)
from unga81.database import create_database, insert_into_country, get_db_connection
from unga81.utils import wordcount
```

```python
# Get list of links to videos
df = get_video_url_list()
```

```python
# Download the transcripts using the speeches urls
download_all_transcripts(df, overwrite=False)
```

```python
create_database()
```

```python
conn = get_db_connection()
for i, r in df.iterrows():
    start, end = get_seconds_from_str(r["start"], r["end"])
    text = get_corpus_from_file(country=r["country"], start=start, end=end)
    insert_into_country(
        conn=conn, row=r, speech=text, wordcount=wordcount(text=text) if text else None
    )


conn.close()
```
