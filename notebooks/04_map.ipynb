```python
# Source - https://stackoverflow.com/a/61129097
# Posted by sentence
# Retrieved 2026-09-24, License - CC BY-SA 4.0

import folium
import json

with open("datasets/world-countries.json") as handle:
    country_geo = json.loads(handle.read())

for i in country_geo["features"]:
    if i["properties"]["name"] == "Germany":
        country = i
        break

m = folium.Map(location=[51.1657, 10.4515], zoom_start=6, min_zoom=5, max_zoom=7)


folium.GeoJson(country, name="germany").add_to(m)

folium.LayerControl().add_to(m)

m
```

```python
import folium

# Initialize the map centered around the coordinates
m = folium.Map(location=[45.5236, -122.6750], zoom_start=13)

# Define the vertices of the polygon [(lat1, lon1), (lat2, lon2), ...]
polygon_coordinates = [
    [45.528, -122.680],
    [45.530, -122.660],
    [45.515, -122.655],
    [45.510, -122.675],
]

# Create and add the polygon to the map
folium.Polygon(
    locations=polygon_coordinates,
    color="blue",  # Outline color
    weight=3,  # Outline thickness
    fill_color="red",  # Fill color
    fill_opacity=0.4,  # Fill opacity
    popup="Portland Area",  # Text on click
    tooltip="Click me!",  # Text on hover
).add_to(m)
m
```
